```python
df_clean = df.copy()

df_clean = df_clean.dropna(
    subset=['play_count', 'digg_count', 'comment_count']
)

print("Jumlah data setelah menghapus missing values:", len(df_clean))
```

```python
df_clean.loc[
    df_clean['duration'] <= 0,
    'duration'
] = np.nan
```

```python
df_clean = df_clean.drop(columns=['repost_count'])
```

```python
df_clean['description'] = df_clean['description'].fillna('No Caption')
```

```python
df_clean['location_created'] = (
    df_clean['location_created'].fillna('Unknown')
)
```

```python
df_clean['upload_datetime'] = pd.to_datetime(
    df_clean['create_time'],
    unit='s',
    errors='coerce'
)

df_clean.loc[
    df_clean['create_time'] <= 0,
    'upload_datetime'
] = pd.NaT
```

# **Membuat Kolom Analisis Marketing**

```python
# Total Engagement
df_clean['total_engagement'] = (
    df_clean['digg_count'] +
    df_clean['comment_count'] +
    df_clean['share_count'] +
    df_clean['collect_count']
)

# Engagement Rate
df_clean['engagement_rate'] = (
    df_clean['total_engagement'] /
    df_clean['play_count']
) * 100

# Like Rate
df_clean['like_rate'] = (
    df_clean['digg_count'] /
    df_clean['play_count']
) * 100

# Comment Rate
df_clean['comment_rate'] = (
    df_clean['comment_count'] /
    df_clean['play_count']
) * 100

# Share Rate
df_clean['share_rate'] = (
    df_clean['share_count'] /
    df_clean['play_count']
) * 100

# Save Rate
df_clean['save_rate'] = (
    df_clean['collect_count'] /
    df_clean['play_count']
) * 100
```

```python
df_clean['upload_year'] = (
    df_clean['upload_datetime'].dt.year
)

df_clean['upload_month'] = (
    df_clean['upload_datetime'].dt.month
)

df_clean['upload_day'] = (
    df_clean['upload_datetime'].dt.day_name()
)

df_clean['upload_hour'] = (
    df_clean['upload_datetime'].dt.hour
)
```

```python
# Convert UTC to WIB
df_clean['upload_datetime'] = (
    pd.to_datetime(
        df_clean['create_time'],
        unit='s',
        utc=True
    ).dt.tz_convert('Asia/Jakarta')
)

# Extract local time features
df_clean['upload_year'] = df_clean['upload_datetime'].dt.year
df_clean['upload_month'] = df_clean['upload_datetime'].dt.month
df_clean['upload_day'] = df_clean['upload_datetime'].dt.day_name()
df_clean['upload_hour'] = df_clean['upload_datetime'].dt.hour

df_clean[['upload_datetime', 'upload_day', 'upload_hour']].head()
```

```python
day_analysis = (
    df_clean.groupby('upload_day')
    .agg(
        total_videos=('video_id', 'count'),
        average_views=('play_count', 'mean'),
        median_views=('play_count', 'median'),
        average_engagement=('engagement_rate', 'mean'),
        median_engagement=('engagement_rate', 'median')
    )
    .reindex([
        'Monday', 'Tuesday', 'Wednesday',
        'Thursday', 'Friday', 'Saturday', 'Sunday'
    ])
)

display(day_analysis.round(2))
```

# **SQL Analysis**

```python
import sqlite3

conn = sqlite3.connect('tiktok_marketing.db')
```

```python
df_clean.to_sql(
    'tiktok_videos',
    conn,
    if_exists='replace',
    index=False
)
```

```python
query = """
SELECT *
FROM tiktok_videos
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    COUNT(*) AS total_videos,
    SUM(play_count) AS total_views,
    SUM(digg_count) AS total_likes,
    SUM(comment_count) AS total_comments,
    SUM(share_count) AS total_shares,
    SUM(collect_count) AS total_saves
FROM tiktok_videos;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    author_unique_id,
    COUNT(*) AS total_videos,
    SUM(play_count) AS total_views,
    ROUND(AVG(play_count), 2) AS average_views,
    ROUND(AVG(engagement_rate), 2) AS average_engagement_rate
FROM tiktok_videos
GROUP BY author_unique_id
ORDER BY average_engagement_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    upload_day,
    COUNT(*) AS total_videos,
    ROUND(AVG(play_count), 2) AS average_views,
    ROUND(AVG(engagement_rate), 2) AS average_engagement_rate
FROM tiktok_videos
GROUP BY upload_day
ORDER BY average_views DESC;
"""

pd.read_sql_query(query, conn)
```
